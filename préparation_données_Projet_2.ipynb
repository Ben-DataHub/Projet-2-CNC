```python

import pandas as pd

lien_basics = "https://datasets.imdbws.com/title.basics.tsv.gz"

colonnes = [
    "tconst",
    "titleType",
    "primaryTitle",
    "isAdult",
    "startYear",
    "runtimeMinutes",
    "genres"
]

basics = pd.read_csv(
    lien_basics,
    sep="\t",
    na_values="\\N",
    usecols=colonnes,
    low_memory=False
)

print("Chargement terminé !")
print("Dimensions :", basics.shape)

basics.head()

```

```python

import pandas as pd

# Charger TMDB depuis ton dossier Projet 2
fichier_cinema = pd.read_csv(
    "TMDB_movie_dataset_v11.csv"
)

# Vérifier les dimensions
print(fichier_cinema.shape)

# Afficher les premières lignes
fichier_cinema.head()
```

```python

# Nombre total de films
print("Total :", len(fichier_cinema))

# Films avec une note >= 7.5
print(
    "Note >= 7,5 :",
    len(fichier_cinema.loc[
        fichier_cinema["vote_average"] >= 7.5
    ])
)

# Films avec au moins 3 000 votes
print(
    "Votes >= 3000 :",
    len(fichier_cinema.loc[
        fichier_cinema["vote_count"] >= 3000
    ])
)

# Films qui respectent les deux conditions
print(
    "Les deux conditions :",
    len(fichier_cinema.loc[
        (fichier_cinema["vote_average"] >= 7.5) &
        (fichier_cinema["vote_count"] >= 3000)
    ])
)
```

```python

comparaison = []

for note in [6.5, 7, 7.5, 8]:
    for votes in [1000, 3000, 5000, 10000]:

        nombre = len(fichier_cinema.loc[
            (fichier_cinema["vote_average"] >= note) &
            (fichier_cinema["vote_count"] >= votes)
        ])

        comparaison.append({
            "Note minimale": note,
            "Votes minimum": votes,
            "Nombre de films": nombre
        })

df_comparaison = pd.DataFrame(comparaison)

df_comparaison
```

```python

print("Nombre de titres IMDb :", len(basics))

print("Nombre de lignes TMDB :", len(fichier_cinema))

print(
    "Nombre de films IMDb :",
    len(basics[basics["titleType"] == "movie"])
)
```

```python

# Préparer les données IMDb
df_films = basics.copy()

# Convertir les colonnes numériques
df_films["startYear"] = pd.to_numeric(
    df_films["startYear"], errors="coerce"
)

df_films["runtimeMinutes"] = pd.to_numeric(
    df_films["runtimeMinutes"], errors="coerce"
)

# Garder les films correspondant au projet
df_films = df_films.loc[
    (df_films["titleType"] == "movie") &
    (df_films["startYear"].between(1920, 2026)) &
    (df_films["runtimeMinutes"] >= 60)
].copy()

print("Nombre de films conservés :", len(df_films))
```

```python

lien_ratings = "https://datasets.imdbws.com/title.ratings.tsv.gz"

ratings = pd.read_csv(
    lien_ratings,
    sep="\t",
    na_values="\\N"
)

ratings.head()
```

```python

df_films_notes = pd.merge(
    df_films,
    ratings,
    on="tconst",
    how="left",
    validate="one_to_one"
)

print("Films après jointure :", len(df_films_notes))
df_films_notes.head()
```

```python

tmdb_unique = fichier_cinema.dropna(
    subset=["imdb_id"]
).drop_duplicates(
    subset=["imdb_id"]
)

df_catalogue = pd.merge(
    df_films_notes,
    tmdb_unique,
    left_on="tconst",
    right_on="imdb_id",
    how="left",
    validate="one_to_one",
    suffixes=("_imdb", "_tmdb")
)

print("Films après jointure :", len(df_catalogue))
print("Films avec correspondance TMDB :",
      df_catalogue["imdb_id"].notna().sum())

df_catalogue
```

```python

print("Films avant jointure :", len(df_films_notes))
print("Films après jointure :", len(df_catalogue))

print("Doublons :", df_catalogue["tconst"].duplicated().sum())

print("Films avec résumé :",
      df_catalogue["overview"].notna().sum())

print("Films avec affiche :",
      df_catalogue["poster_path"].notna().sum())
```

```python

# Convertir les colonnes en numérique
df_catalogue["startYear"] = pd.to_numeric(
    df_catalogue["startYear"], errors="coerce"
)

df_catalogue["isAdult"] = pd.to_numeric(
    df_catalogue["isAdult"], errors="coerce"
)

# Appliquer les 4 filtres
df_films_filtres = df_catalogue.loc[
    (df_catalogue["numVotes"] >= 5000) &
    (df_catalogue["averageRating"] >= 7.5) &
    (df_catalogue["startYear"] >= 1920) &
    (df_catalogue["isAdult"] == 0)
].copy()

print("Nombre de films conservés :", len(df_films_filtres))

df_films_filtres.head()
```

```python

print("Films avant filtrage :", len(df_catalogue))
print("Films après filtrage :", len(df_films_filtres))
print("Nombre de doublons :", df_films_filtres["tconst"].duplicated().sum())
```
